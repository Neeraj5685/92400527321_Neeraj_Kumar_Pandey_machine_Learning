```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler,MinMaxScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest,f_regression
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv('auto-mpg.csv')
df
```

```python
print(df.dtypes)
```

```python
df.select_dtypes(include=np.number).columns
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
print("Numberic Columns:",numeric_columns)
```

```python
categorical_columns = df.select_dtypes(include='object').columns
print("Categorical Columns:",categorical_columns)
```

```python
df["origin"]=df["origin"].astype("object")
```

```python
print(df.isnull().sum())
```

```python
df=df.replace("?",np.nan)
df.isnull().sum()
```

```python
df['horsepower'] = pd.to_numeric(df['horsepower'])
```

```python
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mean())
```

```python
print(df.isnull().sum())
```

```python
scaler=MinMaxScaler().set_output(transform="pandas")
scaled_df=scaler.fit_transform(df[numeric_columns])
scaled_df.head()
```

```python
categorical_columns=["origin","car name"]
```

```python
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
).set_output(transform="pandas")

encoder_data = encoder.fit_transform(df[categorical_columns])

display(encoder_data.head())
```

```python
X = df.drop("mpg",axis=1)
y=df["mpg"]
```

```python
X_numeric = X.select_dtypes(include=np.number)
selector = SelectKBest(
    score_func=f_regression,
    k=3
).set_output(transform='pandas')

X_selected = selector.fit_transform(X_numeric,y)

print(X_selected.head())
```
